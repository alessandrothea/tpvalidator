# VD Geometry — TPC volume Viewer
Displays the TPC volumes from the DUNE geometry JSON using matplotlib voxels.

```python
%load_ext autoreload
%autoreload 2

# %matplotlib widget
```

```python
import matplotlib.pyplot as plt
from pathlib import Path
from tpvalidator.viz.geo_dg import plot_geometry_3d, plot_geometry_3d, animate_tpc_volumes, compare_cryostats_3d
from tpvalidator.detgeometry import load_geodata
import json
from importlib.resources import files


figsize=(12,6)
GEO_FILE = Path('../../data/vd/geo/dunevd10kt_1x8x14_3view_30deg_geo.json')
```

```python
# GEO_FILE = Path('../../../../data/vd/geo/dunevd10kt_1x8x14_3view_30deg_geo.json')
# with open(GEO_FILE) as f:
    # geo = json.load(f)

geodata_1x8x14 = load_geodata('dunevd10kt_1x8x14_3view_30deg_geo.json')

show_beam=False
show_drift=False
show_cryostat_dimensions=False
annotations_overrides={'beam_anchor_y': 650, 'drift_anchor_z': 2100}

fig, ax = plot_geometry_3d(
    geodata_1x8x14, 
    show_cryostat_dimensions=show_cryostat_dimensions,
    # show_axes=False,
    # show_tpcs=False,
    show_opdets=False,
    show_beam=show_beam,
    show_drift=show_drift,
    show_grid=False,
    figsize=figsize,
    annotation_overrides=annotations_overrides
    )
fig, ax = plot_geometry_3d(
    geodata_1x8x14, 
    show_cryostat_dimensions=show_cryostat_dimensions,
    # show_axes=False,
    show_tpcs=False,
    # show_opdets=False,
    show_beam=show_beam,
    show_drift=show_drift,
    show_grid=False,
    figsize=figsize,
    annotation_overrides=annotations_overrides
    )
# plt.show()
```

```python
# GEO_FILE = Path('../../data/vd/geo/dunevd10kt_1x8x6_3view_30deg_geo.json')
fig, ax = plot_geometry_3d(
    geodata_1x8x14, 
    # show_cryostat_dimensions=True,
    # show_axes=False,
    show_beam=True,
    show_drift=True
    )
plt.show()
```
